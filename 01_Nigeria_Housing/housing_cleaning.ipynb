```python
import pandas as pd

# Load the data
df = pd.read_csv('Nigeria Housing Properties Dataset.csv', encoding='ISO-8859-1')

# Show the first 5 rows so I can see what I am working with
df.head()
```

```python
df.info()
```

```python
df['Currency'].value_counts()
```

```python
df['Price']=df['Price'].str.replace(r'[,]','', regex=True).astype(float)
```

```python
df.info()
```

```python
df.head()
```

```python
dollar_row = df['Currency']== '$'

df.loc[dollar_row, 'Price']=df.loc[dollar_row, 'Price']*1500
```

```python
df.head()
```

```python
# Check the head of rows that are still labeled '$'
print(df.loc[df['Currency'] == '$', ['Currency', 'Price']].head())
```

```python
df['Currency']='Naira'
```

```python
df.columns
```

```python
df.drop(columns='Contact', inplace=True)
```

```python
df.duplicated().sum()
```

```python
df.drop_duplicates(inplace=True)
df.shape
```

```python
print(df.isnull().sum()/len(df)*100)
```

```python
df.isnull().sum()
```

```python
df.dropna(subset=['Bedrooms'], inplace=True)
df.info()
```

```python
df.drop(columns=['Parking Spaces'], inplace=True)
```

```python
df.isnull().sum()
```

```python
df.groupby('Bedrooms')['Bathrooms'].median()
```

```python
fill_value = df.groupby('Bedrooms')['Bathrooms'].transform('median')
df['Bathrooms']= df['Bathrooms'].fillna(fill_value)
df.info()
```

```python
df['Toilets'] = df['Toilets'].fillna(df.groupby('Bathrooms')['Toilets'].transform('median'))
df.info()
```

```python
df.head()
df.isnull().sum()
```

```python
df.dropna(inplace=True)
df.info()
```

```python
import matplotlib.pyplot as plt
plt.hist(df['Price'], bins=50);
```

```python
pd.set_option('display.float_format', lambda x: '%.2f' % x) # This turns off scientific notation
print(df['Price'].describe())
```

```python
df['Price'].quantile(0.95)
```

```python
plt.hist(df[df['Price']< 1211430000]['Price'], bins= 50)
plt.title('Distribution of House Prices (< 95th Percentile)')
plt.xlabel('Price [Naira]')
plt.ylabel('Frequency')
plt.show()
```

```python
df['Location'].value_counts()
```

```python
top_10 = df['Location'].value_counts().head(10)
plt.figure(figsize=(15, 7))
plt.xticks(rotation=60)
plt.bar(top_10.index, height=top_10.values)
```

```python
top_10_names = df['Location'].value_counts().head(10).index
top_10_names
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Make the "VIP List" of the top 10 most popular locations
top_10_locations = df['Location'].value_counts().head(10).index

# 2. Create a clean mini-table with ONLY those locations AND normal prices (< 1.2 Billion)
df_clean = df[(df['Location'].isin(top_10_locations)) & (df['Price'] < 1211430000)]

# 3. Draw the Box Plot!
plt.figure(figsize=(12, 6))
sns.boxplot(data=df_clean, x='Location', y='Price')

# Make the text look nice
plt.xticks(rotation=45)
plt.title('House Prices in Top 10 Locations (Excluding Billionaire Outliers)')
plt.ylabel('Price (Naira)')
plt.show()
```

```python
top_10_states = df['State'].value_counts().head(10).index
df_clean_state = df[(df['State'].isin(top_10_states)) & (df['Price'] < 1211430000)]
plt.Figure(figsize=(10, 6))
sns.boxplot(data=df_clean_state, x='State', y='Price')
plt.xticks(rotation =46)
plt.title('House prices in Top 10 States');
```

```python
num_col = ['Price', 'Bedrooms', 'Bathrooms', 'Toilets']

df[num_col].corr().round(2)
```

```python
# Use the exact same logic, just change df to df_clean
numerical_cols = ['Price', 'Bedrooms', 'Bathrooms', 'Toilets']
correlation_matrix = df_clean[numerical_cols].corr()
print(correlation_matrix)
```

```python
plt.Figure(figsize=(8,6))
sns.heatmap(data=correlation_matrix, annot=True, cmap='Greens');
```
